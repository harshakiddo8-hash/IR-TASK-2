```python

# ==========================================================
# TEXT CLASSIFICATION COMPARATIVE STUDY
# Decision Tree vs KNN vs SVM vs Naive Bayes
# ==========================================================

# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import urllib.request
import os

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# 2. DOWNLOAD DATASET

url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"

urllib.request.urlretrieve(url, "sms_spam.zip")

with zipfile.ZipFile("sms_spam.zip", "r") as zip_ref:
    zip_ref.extractall("sms_dataset")

file_path = "sms_dataset/SMSSpamCollection"

df = pd.read_csv(
    file_path,
    sep="\t",
    names=["Label", "Message"],
    encoding="utf-8"
)

print("DATASET INFORMATION")
print(df.head())
print("\nDataset Shape:", df.shape)
print("\nClass Distribution:")
print(df["Label"].value_counts())

# 3. DATA PREPROCESSING

df.dropna(inplace=True)
df.drop_duplicates(inplace=True)

df["Label"] = df["Label"].str.lower().str.strip()
df["Message"] = df["Message"].astype(str).str.lower()

df = df[df["Label"].isin(["ham", "spam"])]

print("\nCleaned Dataset:")
print(df.head())

# 4. ENCODE CLASS LABELS

df["Label_Number"] = df["Label"].map({
    "ham": 0,
    "spam": 1
})

X = df["Message"]
y = df["Label_Number"]

# 5. TRAIN TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 6. TF-IDF FEATURE EXTRACTION

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=3000,
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("\nTF-IDF FEATURES:", X_train_tfidf.shape[1])
print("Training Samples:", X_train_tfidf.shape[0])
print("Testing Samples:", X_test_tfidf.shape[0])

# 7. INITIALIZE FOUR CLASSIFIERS

models = {
    "Decision Tree": DecisionTreeClassifier(
        random_state=42,
        max_depth=100
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5,
        metric="cosine",
        algorithm="brute"
    ),

    "SVM": LinearSVC(
        C=1.0,
        random_state=42,
        max_iter=5000
    ),

    "Naive Bayes": MultinomialNB(
        alpha=1.0
    )
}

# 8. TRAIN AND EVALUATE MODELS

results = []
predictions = {}

for name, model in models.items():

    model.fit(X_train_tfidf, y_train)

    y_pred = model.predict(X_test_tfidf)

    predictions[name] = y_pred

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    results.append({
        "Algorithm": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

    print("\n", "=" * 50)
    print(name)
    print("=" * 50)

    print(classification_report(
        y_test,
        y_pred,
        target_names=["Ham", "Spam"],
        zero_division=0
    ))

# 9. COMPARATIVE PERFORMANCE TABLE

results_df = pd.DataFrame(results)

print("\nCOMPARATIVE PERFORMANCE")
display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}"
    })
)

# 10. PERFORMANCE COMPARISON GRAPH

metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]

results_df.set_index("Algorithm")[metrics].plot(
    kind="bar",
    figsize=(12, 6),
    ylim=(0, 1.05),
    width=0.8
)

plt.title("Comparison of Four Text Classification Algorithms")
plt.xlabel("Algorithms")
plt.ylabel("Performance Score")
plt.xticks(rotation=20)
plt.legend(loc="lower right")
plt.grid(axis="y")
plt.tight_layout()
plt.show()

# 11. CONFUSION MATRIX FOR ALL MODELS

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, (name, y_pred) in zip(
    axes.flatten(), predictions.items()
):

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Ham", "Spam"],
        yticklabels=["Ham", "Spam"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted Label")
    ax.set_ylabel("Actual Label")

plt.tight_layout()
plt.show()

# 12. IDENTIFY BEST MODEL

best_model_name = results_df.loc[
    results_df["F1-Score"].idxmax(),
    "Algorithm"
]

print("\nBEST PERFORMING MODEL:", best_model_name)

print(
    "Best F1 Score:",
    results_df["F1-Score"].max()
)

# 13. WHAT-IF EXPERIMENT
# Change SVM parameter C from 1.0 to 0.1 and 10.0

c_values = [0.1, 1.0, 10.0]

what_if_results = []

for c in c_values:

    svm_model = LinearSVC(
        C=c,
        random_state=42,
        max_iter=5000
    )

    svm_model.fit(X_train_tfidf, y_train)

    pred = svm_model.predict(X_test_tfidf)

    what_if_results.append({
        "SVM C Value": c,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(
            y_test, pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, pred, zero_division=0
        ),
        "F1-Score": f1_score(
            y_test, pred, zero_division=0
        )
    })

what_if_df = pd.DataFrame(what_if_results)

print("\nWHAT-IF EXPERIMENT RESULTS")
display(what_if_df.round(4))

# 14. WHAT-IF GRAPH

what_if_df.set_index("SVM C Value")[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].plot(
    kind="bar",
    figsize=(10, 5),
    ylim=(0, 1.05)
)

plt.title("What-If Experiment: Effect of SVM C Parameter")
plt.xlabel("SVM C Value")
plt.ylabel("Performance Score")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.tight_layout()
plt.show()

# 15. FINAL RECOMMENDATION

best_what_if = what_if_df.loc[
    what_if_df["F1-Score"].idxmax()
]

print("\nFINAL EXPERIMENT SUMMARY")
print("-" * 45)

print("Dataset: SMS Spam Collection")
print("Classification: Ham and Spam")
print("Algorithms: Decision Tree, KNN, SVM, Naive Bayes")
print("Feature Extraction: TF-IDF")
print("Train-Test Ratio: 80:20")

print("\nBest Classifier:", best_model_name)
print("Best F1 Score:", round(
    results_df["F1-Score"].max(), 4
))

print("\nBest SVM C Value:", best_what_if["SVM C Value"])
print("Best SVM F1 Score:", round(
    best_what_if["F1-Score"], 4
))

# 16. SAVE RESULTS

df.to_csv("SMS_Spam_Dataset.csv", index=False)

results_df.to_csv(
    "Classification_Comparison.csv",
    index=False
)

what_if_df.to_csv(
    "What_If_Experiment.csv",
    index=False
)

print("\nALL EXPERIMENTS COMPLETED SUCCESSFULLY!")

# 17. DOWNLOAD RESULT FILES

from google.colab import files

files.download("Classification_Comparison.csv")
```
